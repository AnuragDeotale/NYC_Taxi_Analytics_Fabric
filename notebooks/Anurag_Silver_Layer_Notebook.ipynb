**Import Libraries**

```python
from pyspark.sql.functions import *
```

**Load Bronze Data**

```python
df = spark.read.table("anurag_bronze_trips")
# display(df)
```

```python

total_rows = df.count()
print(total_rows)
```

**"lpepPickupDatetime" Column**

```python
df_clean = df.filter(

    (year(col("lpepPickupDatetime")) >= 2008) &
    (year(col("lpepPickupDatetime")) <= 2018)

)

# display(df_clean)
```

```python
df_clean = df_clean.withColumn(

    "PickUpDate",
    to_date(col("lpepPickupDatetime"))

).withColumn(

    "PickUpTime",
    date_format(col("lpepPickupDatetime"), "HH:mm:ss")

)

# display(df_clean)
```

```python

'''
total_rows = df_clean.count()
print(total_rows)
'''
```

**“lpepDropoffDatetime” Column**

```python
df_clean_2 = df_clean.filter(

    (year(col("lpepPickupDatetime")) >= 2008) &
    (year(col("lpepPickupDatetime")) <= 2018)

)

# display(df_clean)
```

```python
df_clean_2 = df_clean.withColumn(

    "DropOffDate",
    to_date(col("lpepDropoffDatetime"))

).withColumn(

    "DropOffTime",
    date_format(col("lpepDropoffDatetime"), "HH:mm:ss")

)

#display(df_clean_2)
```

```python

'''
total_rows_2= df_clean_2.count()
print(total_rows_2)
'''
```

**"passengerCount" Column**

```python
df_clean_3= df_clean_2.filter(col("passengerCount") > 0)
# display(df_clean_3)
```

```python

'''
total_rows_3 = df_clean_3.count()
print(total_rows_3)
'''
```

**"tripDistance" Column**

```python
df_clean_4= df_clean_3.filter(col("tripDistance") > 0)
# display(df_clean_4)
```

```python

'''
total_rows_4 = df_clean_4.count()
print(total_rows_4)
'''
```

**"rateCodeId" Column**

```python
df_clean_5 = df_clean_4.filter(col("rateCodeID").isin(1,2,3,4,5,6))
#display(df_clean_5)
```

```python

'''
total_rows_5 = df_clean_5.count()
print(total_rows_5)
'''
```

**"storeAndFWdFlag" Column**

```python
df_clean_6 = df_clean_5.withColumn(
    "storeAndFwdFlag",
    
    when(col("storeAndFwdFlag") == "Y", 1)
    .when(col("storeAndFwdFlag") == "N", 0)
)
#display(df_clean_6)
```

```python


total_rows_6= df_clean_6.count()
print(total_rows_6)
```

**"fareAmount" Column**

```python
df_clean_7 = df_clean_6.filter(col("fareAmount") > 0)
#display(df_clean_7)
```

```python
'''
total_rows_7 = df_clean_7.count()
print(total_rows_7)
'''
```

**"extra" Column**

```python
df_clean_8 = df_clean_7.filter(
    col("extra").isin(0, 0.5, 1)
)
#display(df_clean_8)
```

```python
'''
total_rows_8 = df_clean_8.count()
print(total_rows_8)
'''
```

**"mtaTax" Column**

```python
df_clean_9= df_clean_8.filter(
    (col("mtaTax") == 0.5) |
    (col("mtaTax") == 0)
)
#display(df_clean_9)
```

```python
'''
total_rows_9 = df_clean_9.count()
print(total_rows_9)
'''
```

**“improvementSurcharge” Column**

```python
df_clean_10= df_clean_9.withColumn(
    "improvementSurcharge",
    col("improvementSurcharge").cast("float")
)

#display(df_clean_10)
```

```python
'''
total_rows_10 = df_clean_10.count()
print(total_rows_10)
'''
```

```python
'''
df_clean_10.groupBy("improvementSurcharge") \
        .count() \
        .orderBy("improvementSurcharge") \
        .show()
'''
```

```python
df_clean_11 = df_clean_10.withColumn(

    "improvementSurcharge",

    when(
        year(col("lpepPickupDatetime")) < 2015,
        0
    ).otherwise(col("improvementSurcharge"))

)

#display(df_clean_11)
```

```python
'''
df_clean_11.groupBy("improvementSurcharge") \
        .count() \
        .orderBy("improvementSurcharge") \
        .show()
'''
```

```python


df_clean_12 = df_clean_11.withColumn(

    "improvementSurcharge",

    trim(col("improvementSurcharge"))

).withColumn(

    "improvementSurcharge",

    round(
        col("improvementSurcharge").cast("double"),
        2
    )

).filter(

    (col("improvementSurcharge") == 0.0) |
    (col("improvementSurcharge") == 0.3)

)

'''
df_clean_11.groupBy("improvementSurcharge") \
          .count() \
          .orderBy("improvementSurcharge") \
          .show()
'''
# display(df_clean_12)
```

**"tipAmount" Column**

```python
df_clean_13= df_clean_12.filter(
    col("tipAmount") >= 0
)
```

**"tollsAmount" Column**

```python
df_clean_14= df_clean_13.filter(
    col("tollsAmount") >= 0
)
```

**"totalAmount" Column**

```python
df_clean_15= df_clean_14.filter(
    col("totalAmount") >= 0
)
```

```python
'''
total_rows_15= df_clean_15.count()
print(total_rows_15)
'''
```

**"tripType" Column**

```python
df_clean_16= df_clean_15.withColumn(

    "tripType",

    when(col("tripType").isNull(), 1)
    .otherwise(col("tripType"))

)
```

```python
df_clean_16 = df_clean_16.withColumn(

    "tripType",

    when(col("tripType") == 1, 0)
    .when(col("tripType") == 2, 1)

)


#display(df_clean_16)
```

```python
'''
total_rows_16= df_clean_16.count()
print(total_rows_16)
'''
```

**Dropping Columns with High NULL VALUES AND NOT REQUIRED**

```python
df_clean_17= df_clean_16.drop(

    "lpepPickupDatetime",
    "lpepDropoffDatetime",
    "puLocationId",
    "doLocationId",
    "pickupLongitude",
    "pickupLatitude",
    "dropoffLongitude",
    "dropoffLatitude",
    "ehailFee"

)

#display(df_clean_17)
```

```python
'''
total_rows_17= df_clean_17.count()
print(total_rows_17)
'''
```
